```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("used_cars.csv")
df
```

```python
df_lim = df.head(100)
display(df_lim)
```

```python
# Histogram (Numerical)

df_lim.loc[:, 'price'] = df_lim['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
df_lim.loc[:, 'price'] = pd.to_numeric(df_lim['price'], errors='coerce')

df_lim_for_hist = df_lim.dropna(subset=['price'])

plt.hist(df_lim_for_hist.price, rwidth=0.8, bins=20)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution of Price")
plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

```python
# Pie (Categorical)

fuelT_counts = df_lim['fuel_type'].value_counts()
plt.pie(fuelT_counts, labels = fuelT_counts.index, autopct='%1.1f%%')
plt.title("Fuel Type")
plt.show()
```

```python
# Bar (Categorical)
transmission_counts = df_lim['transmission'].value_counts()
plt.bar(transmission_counts.index, transmission_counts.values)
plt.xlabel("Transmission Type")
plt.ylabel("Count")
plt.title("Distribution of Transmission Types")
plt.xticks(rotation=90)
plt.show()
```

```python
# Scatter Plot (Numerical x Numerical)

df_lim['price'] = df_lim['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
df_lim['price'] = pd.to_numeric(df_lim['price'], errors='coerce')

df_lim['milage'] = df_lim['milage'].astype(str).str.replace(' mi.', '', regex=False).str.replace(',', '', regex=False)
df_lim['milage'] = pd.to_numeric(df_lim['milage'], errors='coerce')
df_lim_for_plot = df_lim.dropna(subset=['price', 'milage'])

plt.scatter(df_lim_for_plot.milage , df_lim_for_plot.price , marker = '.')
plt.xlabel("Milage")
plt.ylabel("Price")
plt.title("Milage vs Price")
plt.show()
```

```python
avg_price_by_brand = df_lim.groupby('brand')['price'].mean().reset_index()

plt.figure(figsize=(12, 5))
plt.bar(avg_price_by_brand.brand, avg_price_by_brand.price)
plt.xlabel("Brand")
plt.ylabel("Average Price")
plt.title("Average Price by Brand")
plt.xticks(rotation=90)
plt.grid(axis='y')
plt.show()
```
